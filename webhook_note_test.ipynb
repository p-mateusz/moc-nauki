# Test webhooka do notatek

Notebook służy do weryfikacji dostępności webhooka oraz wysyłania przykładowych danych notatki.

```python
import requests
import json

print('requests version:', requests.__version__)
```

```python
webhook_url = 'https://czacki3.app.n8n.cloud/webhook/a5aa889e-719f-44f5-a68f-21eda63cb52c'
print('Webhook URL:', webhook_url)
```

```python
try:
    response = requests.post(webhook_url, json={
        'event': 'test',
        'source': 'notatki-test',
        'content': 'To jest testowy payload webhooka dla notatek.'
    }, timeout=15)
    print('Status code:', response.status_code)
    print('Response headers:', response.headers)
    try:
        print('Response body:', response.text)
    except Exception as e:
        print('Nie można odczytać ciała odpowiedzi:', e)
except Exception as e:
    print('Błąd podczas wysyłania żądania do webhooka:', e)
```

```python
try:
    note_payload = {
        'title': 'Testowa notatka z webhooka',
        'content': 'Przykładowa treść notatki wygenerowana przez webhook.',
        'created_at': '2026-04-04T12:00:00Z',
        'note_type': 'audio',
        'source': 'webhook-test',
    }
    response = requests.post(webhook_url, json=note_payload, timeout=15)
    print('Status code:', response.status_code)
    print('Response body:', response.text)
except Exception as e:
    print('Błąd przy tworzeniu notatki:', e)
```

```python
# Test direct n8n webhook with sample flashcard data
import requests
import json

webhook_url = 'https://czacki3.app.n8n.cloud/webhook-test/7d73923d-05ab-482b-b453-8308ac693381'
payload = {
    'formattedWords': '{hola | cześć}, {adiós | do widzenia}, {amigo | -}',
    'words': '{hola | cześć}, {adiós | do widzenia}, {amigo | -}',
    'language': 'es',
    'definitionLanguage': 'język docelowy (es)',
    'definitionLanguageMode': 'target',
    'specialInstructions': 'Użyj prostych zdań z luką i podaj zdania w języku hiszpańskim. Jeśli definicja jest pusta, użyj myślnika i przejdź dalej.',
}

print('Sending test payload to:', webhook_url)
print('Payload:', json.dumps(payload, ensure_ascii=False, indent=2))

try:
    response = requests.post(webhook_url, json=payload, timeout=30)
    print('Status code:', response.status_code)
    print('Response headers:', response.headers)
    print('Response body:', response.text)
except Exception as e:
    print('Request failed:', e)
```
